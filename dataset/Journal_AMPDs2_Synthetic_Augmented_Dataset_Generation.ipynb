### Unsupervised Code: Benign Data_augmented code

```python
import csv
import random
import os

# Define the path to the directory containing CSV files
data_path = "C:\\Users\\olufe\\projects\\Appliances_Data"

# List of names to process
names = ['B1E', 'B2E', 'BME', 'CDE', 'CWE', 'DNE', 'DWE',
         'EBE', 'EQE', 'FGE', 'FRE', 'GRE', 'HPE', 'HTE',
         'OFE', 'OUE', 'RSE', 'TVE', 'UTE', 'WHE', 'WOE']

# Time and day offsets
time_offset = 7 * 60 * 60
day_offset = 15431

power_data = {}

home_type = 2
train_data = []
test_data = []

train_count = {
    'Benign': 0, 'Baseload': 0, 'Weakload': 0, 'Peakhour': 0,
    'Midnight': 0, 'Eviltwin': 0, 'ImpactSum': 0, 'ImpactCnt': 0, 'LegitimateSum': 0
}
test_count = train_count.copy()

random.seed(1)

# Data loading from the files
for name in names:
    power_data[name] = {}

    filename = os.path.join(data_path, f'Electricity_{name}.csv')
    with open(filename, 'r') as csv_file:
        f = csv.reader(csv_file, delimiter=',')

        maximum = 0
        count = -1
        for row in f:
            count += 1
            if count == 0:
                continue

            time = int(row[0]) - time_offset
            power = int(row[6])
            if maximum < power:
                maximum = power

            day = int(time / 86400) - day_offset
            minute = int((time % 86400) / 60)

            if day not in power_data[name]:
                power_data[name][day] = {}

            if minute not in power_data[name][day]:
                power_data[name][day][minute] = power

        print(f'{name}: {maximum}')

# Data check
for name in names:
    for day in range(730):
        if day in power_data[name]:
            for minute in range(1440):
                if minute not in power_data[name][day]:
                    print(f'{name}: minute={minute} was not found on day={day}')
        else:
            print(f'{name}: day={day} was not found')

# Aggregate Data
power_base = {}
for day in range(730):
    power_base[day] = {}
    for minute in range(1440):
        power_base[day][minute] = 0

base_names = ['B1E', 'B2E', 'BME', 'CWE', 'DNE', 'HTE',
              'EBE', 'EQE', 'FRE', 'OFE', 'OUE', 'TVE', 'UTE',
              'CDE', 'HPE', 'DWE', 'FGE', 'WOE']

home_name = 'A'
if home_type == 1:
    home_name = 'B'
    base_names.remove('CDE')
    base_names.remove('WOE')
elif home_type == 2:
    home_name = 'C'
    base_names.remove('DWE')
    base_names.remove('HPE')
    base_names.remove('TVE')

for name in base_names:
    for day in range(730):
        for minute in range(1440):
            power_base[day][minute] += power_data[name][day][minute]

train_split = int(730 * 0.8)

for day in range(730):
    legitimate_sum = 0
    row = []
    for minute in range(1440):
        row.append(power_base[day][minute])
        legitimate_sum += power_base[day][minute]
    row.append('0')
    if day < train_split:
        train_data.append(row)
    else:
        test_data.append(row)

train_data_aug = []
for d in train_data:
    for starting_minute in range(60):
        row = []
        for minute in range(1440 - 60):
            row.append(d[starting_minute + minute])
        row.append(d[-1])
        train_data_aug.append(row)

test_data_aug = []
for d in test_data:
    row = []
    for minute in range(1440 - 60):
        row.append(d[minute])
    row.append(d[-1])
    test_data_aug.append(row)

train_attack_data_aug = []
for d in train_data_aug:
    label = random.randint(1, 5)
    impact_sum = 0
    row = []

    if label == 1:  # baseload attack
        for minute in range(1440 - 60):
            attack = 100
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 2:  # weakload attack
        for minute in range(1440 - 60):
            attack = 0
            if d[minute] >= 500:
                attack = 500
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 3:  # peakhour attack
        for minute in range(1440 - 60):
            attack = 0
            if d[minute] >= 1500:
                attack = 2000
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 4:  # midnight attack
        starting_time = random.randint(60, 60 + 60)
        attack_duration = 120
        for minute in range(1440 - 60):
            attack = 0
            if starting_time <= minute and minute <= starting_time + attack_duration:
                attack = 1000
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 5:  # evil-twin attack
        evil_twin_day = random.randrange(0, len(train_data_aug))
        for minute in range(1440 - 60):
            attack = train_data_aug[evil_twin_day][minute]
            row.append(d[minute] + attack)
            impact_sum += attack

    row.append('1')
    if impact_sum > 0:
        train_attack_data_aug.append(row)

train_output = []
train_output.extend(train_data_aug)
random.shuffle(train_output)

test_attack_data_aug = []
for d in test_data_aug:
    label = random.randint(1, 5)
    impact_sum = 0
    row = []

    if label == 1:  # baseload attack
        for minute in range(1440 - 60):
            attack = 100
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 2:  # weakload attack
        for minute in range(1440 - 60):
            attack = 0
            if d[minute] >= 500:
                attack = 500
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 3:  # peakhour attack
        for minute in range(1440 - 60):
            attack = 0
            if d[minute] >= 1500:
                attack = 2000
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 4:  # midnight attack
        starting_time = random.randint(60, 60 + 60)
        attack_duration = 120
        for minute in range(1440 - 60):
            attack = 0
            if starting_time <= minute and minute <= starting_time + attack_duration:
                attack = 1000
            row.append(d[minute] + attack)
            impact_sum += attack
    elif label == 5:  # evil-twin attack
        evil_twin_day = random.randrange(0, len(test_data_aug))
        for minute in range(1440 - 60):
            attack = test_data_aug[evil_twin_day][minute]
            row.append(d[minute] + attack)
            impact_sum += attack

    row.append('1')
    if impact_sum > 0:
        test_attack_data_aug.append(row)

test_output = []
test_output.extend(test_data_aug)
test_output.extend(test_attack_data_aug)
random.shuffle(test_output)

csv_header = []
for i in range(1440 - 60):
    csv_header.append('M' + str(i))
csv_header.append('Label')

################################################
# Specify the directory path
output_directory = 'dataset'

# Check if the directory exists, and create it if it doesn't
if not os.path.exists(output_directory):
    os.makedirs(output_directory)

# Specify the file path
output_file_path = os.path.join(output_directory, f'psa_journal_home_{home_name}_unsupervise_train.csv')

# Open the file for writing
with open(output_file_path, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(csv_header)
    writer.writerows(train_output)
###############################################

# f = open(f'dataset/psa_journal_home_{home_name}_unsupervise_train.csv', 'w')
# writer = csv.writer(f)
# writer.writerow(csv_header)
# writer.writerows(train_output)
# f.close()
```
