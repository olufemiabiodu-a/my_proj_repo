```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Paths to the CSV files
files = {
    "Q": "C:\\Users\\olufe\\projects\\Appliances_Data\\Q.csv",
    "P": "C:\\Users\\olufe\\projects\\Appliances_Data\\P.csv",
    "I": "C:\\Users\\olufe\\projects\\Appliances_Data\\I.csv",
    "S": "C:\\Users\\olufe\\projects\\Appliances_Data\\S.csv"
}

# Initialize a dictionary to store correlation values
correlation_values = {}

# Process each file separately
for feature_name, filepath in files.items():
    df = pd.read_csv(filepath)

    # Assuming 'Label' is the last column
    correlation = df.iloc[:, :-1].apply(lambda x: x.corr(df['Label']))
    correlation_values[feature_name] = correlation.mean()

# Convert the correlation values to a DataFrame for plotting
correlation_df = pd.DataFrame.from_dict(correlation_values, orient='index', columns=['Correlation'])

# Plotting the correlations using a bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=correlation_df.index, y=correlation_df['Correlation'])
plt.title('Feature Correlation with Label')
plt.xlabel('Features')
plt.ylabel('Correlation Coefficient')
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

# Load the dataset
# data = pd.read_csv('path_to_your_file.csv')
data = pd.read_csv(r"C:\Users\olufe\projects\Journal\dataset\psa_journal_home_A_synthetic_train.csv")

# Handle missing values (if any)
# data.fillna(method='ffill', inplace=True) # or another appropriate method

# Encode categorical variables if necessary
# If 'label' is categorical:
# label_encoder = LabelEncoder()
# data['label'] = label_encoder.fit_transform(data['label'])

# Define features and target
X = data.drop('Label', axis=1)  # assuming 'label' is the name of your target variable
y = data['Label']

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Mutual Information
mi_scores = mutual_info_classif(X_train, y_train)
mi_scores = pd.Series(mi_scores, name="MI Scores", index=X_train.columns)

# Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
rf_importances = pd.Series(rf.feature_importances_, index=X_train.columns)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
mi_scores.sort_values(ascending=False).plot.bar(ax=axes[0])
axes[0].set_title("Mutual Information Scores")
rf_importances.sort_values(ascending=False).plot.bar(ax=axes[1])
axes[1].set_title("Random Forest Feature Importances")
plt.tight_layout()
plt.show()
```
