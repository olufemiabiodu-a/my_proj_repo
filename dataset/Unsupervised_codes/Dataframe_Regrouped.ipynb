```python
import pandas as pd

# Read the CSV file
csv_path = r'C:\Users\olufe\projects\Journal\dataset\HomeA_unsupervised_sim_combined_data.csv'
df = pd.read_csv(csv_path)

# Select the 1430 columns as X coordinates
x_columns = df.columns[:1380]  # Assuming the first 1380 columns are the X coordinates

# Select the desired column as Y coordinate
y_column = 'Label'  # Update with your desired Y coordinate column name

# Create a new dataframe with the selected columns
selected_df = df[x_columns.append(pd.Index([y_column]))]

# Sort the data by X coordinates
selected_df.sort_values(list(x_columns), inplace=True)

# Initialize a list to store the regrouped points
regrouped_points = []

# Iterate over the rows of the dataframe
for index, row in selected_df.iterrows():
    if not regrouped_points:
        # If the list is empty, add the first point
        regrouped_points.append(row)
    else:
        # Get the last point in the regrouped list
        last_point = regrouped_points[-1]

        # Check if the current point overlaps with the last point
        if all(row[x] <= last_point[x] for x in x_columns) and row[y_column] <= last_point[y_column]:
            # Update the last point's coordinates if the current point overlaps
            last_point.update(row)
        else:
            # Add the current point to the regrouped list if it doesn't overlap
            regrouped_points.append(row)

# Create a new dataframe from the regrouped points list
regrouped_df = pd.DataFrame(regrouped_points)

# Save the regrouped data to a new CSV file
regrouped_csv_path = r'C:\Users\olufe\projects\Journal\dataset\HomeA_regrouped_data.csv'
regrouped_df.to_csv(regrouped_csv_path, index=False)

print("Regrouping completed and saved to", regrouped_csv_path)
```
