```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix

# Load the train and test datasets
train_data_path = "C:/Users/olufe/projects/Journal/dataset/psa_journal_home_A_unsupervise_train.csv"
test_data_path = "C:/Users/olufe/projects/Journal/dataset/psa_journal_home_A_sim_test.csv"

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Convert data to numpy arrays and normalize the data to float32
X_train = train_df.values.astype('float32')
X_test = test_df.values.astype('float32')

# Normalize the data to have values between 0 and 1
X_train = (X_train - X_train.min()) / (X_train.max() - X_train.min())
X_test = (X_test - X_test.min()) / (X_test.max() - X_test.min())
```

```python
# Define the MLP Autoencoder model
input_dim = X_train.shape[1]
encoded_dim = int(input_dim / 2)

model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(input_dim,)),
    keras.layers.Dense(encoded_dim, activation='relu'),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(input_dim, activation='sigmoid')
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='mse')

# Function to calculate performance metrics
def calculate_metrics(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    tnr = tn / (tn + fp)
    fpr = fp / (tn + fp)
    fnr = fn / (fn + tp)
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1 = 2 * (precision * recall) / (precision + recall)
    auc = roc_auc_score(y_true, y_pred)
    return tnr, fpr, fnr, precision, recall, f1, auc
```

```python
# Perform k-fold cross-validation
kfold = KFold(n_splits=3, shuffle=True, random_state=42)

all_metrics = []
for train_idx, test_idx in kfold.split(X_train):
    X_train_fold, X_val_fold = X_train[train_idx], X_train[test_idx]

    # Train the model
    model.fit(X_train_fold, X_train_fold, epochs=50, batch_size=32, verbose=0)

    # Make predictions on the validation set
    val_predictions = model.predict(X_val_fold)

    # Binarize predictions using a threshold (e.g., 0.5)
    # threshold = 0.5
    threshold = 0
    val_predictions_binarized = (val_predictions > threshold).astype(int)

    # Calculate performance metrics for this fold
    tnr, fpr, fnr, precision, recall, f1, auc = calculate_metrics(X_val_fold.flatten(), val_predictions_binarized.flatten())

    all_metrics.append((tnr, fpr, fnr, precision, recall, f1, auc))

# Calculate average performance metrics over all folds
mean_metrics = np.mean(all_metrics, axis=0)

# Print the results
print("Average TNR: ", mean_metrics[0])
print("Average FPR: ", mean_metrics[1])
print("Average FNR: ", mean_metrics[2])
print("Average Precision: ", mean_metrics[3])
print("Average Recall: ", mean_metrics[4])
print("Average F1-score: ", mean_metrics[5])
print("Average AUC: ", mean_metrics[6])
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix

# Load the train and test datasets
train_data_path = "C:/Users/olufe/projects/Journal/dataset/psa_journal_home_A_unsupervise_train.csv"
test_data_path = "C:/Users/olufe/projects/Journal/dataset/psa_journal_home_A_sim_test.csv"

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Convert data to numpy arrays and normalize the data to float32
X_train = train_df.values.astype('float32')
X_test = test_df.values.astype('float32')

# Normalize the data to have values between 0 and 1
X_train = (X_train - X_train.min()) / (X_train.max() - X_train.min())
X_test = (X_test - X_test.min()) / (X_test.max() - X_test.min())

# Define the MLP Autoencoder model
input_dim = X_train.shape[1]
encoded_dim = int(input_dim / 2)

model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(input_dim,)),
    keras.layers.Dense(encoded_dim, activation='relu'),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(input_dim, activation='sigmoid')
])

# Compile the model
model.compile(optimizer='adam', loss='mse')

# Function to calculate performance metrics
def calculate_metrics(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    tnr = tn / (tn + fp)
    fpr = fp / (tn + fp)
    fnr = fn / (fn + tp)
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1 = 2 * (precision * recall) / (precision + recall)
    auc = roc_auc_score(y_true, y_pred)
    return tnr, fpr, fnr, precision, recall, f1, auc

# Function to determine the anomaly threshold from the reconstruction error
def determine_anomaly_threshold(X_train, X_val, threshold_factor=1.0):
    reconstruction_errors = np.mean(np.square(X_val - X_train), axis=1)
    anomaly_threshold = np.mean(reconstruction_errors) + threshold_factor * np.std(reconstruction_errors)
    return anomaly_threshold

# Perform k-fold cross-validation
kfold = KFold(n_splits=5, shuffle=True, random_state=42)

all_metrics = []
for train_idx, val_idx in kfold.split(X_train):
    X_train_fold, X_val_fold = X_train[train_idx], X_train[val_idx]

    # Train the model
    model.fit(X_train_fold, X_train_fold, epochs=50, batch_size=32, verbose=0)

    # Make predictions on the validation set
    val_predictions = model.predict(X_val_fold)

    # Determine the anomaly threshold from the reconstruction error on the training set
    anomaly_threshold = determine_anomaly_threshold(X_train_fold, val_predictions)

    # Binarize predictions based on the anomaly threshold
    val_predictions_binarized = (np.mean(np.square(val_predictions - X_val_fold), axis=1) > anomaly_threshold).astype(int)

    # Calculate performance metrics for this fold
    tnr, fpr, fnr, precision, recall, f1, auc = calculate_metrics(X_val_fold.flatten(), val_predictions_binarized.flatten())

    all_metrics.append((tnr, fpr, fnr, precision, recall, f1, auc))

# Calculate average performance metrics over all folds
mean_metrics = np.mean(all_metrics, axis=0)

# Print the results
print("Average TNR: ", mean_metrics[0])
print("Average FPR: ", mean_metrics[1])
print("Average FNR: ", mean_metrics[2])
print("Average Precision: ", mean_metrics[3])
print("Average Recall: ", mean_metrics[4])
print("Average F1-score: ", mean_metrics[5])
print("Average AUC: ", mean_metrics[6])
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import KFold
from sklearn.metrics import roc_auc_score, confusion_matrix

# Load the dataset
data_path = "C:/Users/olufe/projects/Journal/dataset/HomeA_unsupervised_sim_combined_shuffled.csv"
data_df = pd.read_csv(data_path)

# Convert data to numpy arrays and normalize the data to float32
data = data_df.values.astype('float32')
data = (data - data.min()) / (data.max() - data.min())

# Define the MLP Autoencoder model
input_dim = data.shape[1]
encoded_dim = int(input_dim / 2)

model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(input_dim,)),
    keras.layers.Dense(encoded_dim, activation='relu'),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(input_dim, activation='sigmoid')
])
# Compile the model
model.compile(optimizer='adam', loss='mse')

# Function to calculate performance metrics
def calculate_metrics(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    tnr = tn / (tn + fp)
    fpr = fp / (tn + fp)
    fnr = fn / (fn + tp)
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1 = 2 * (precision * recall) / (precision + recall)
    auc = roc_auc_score(y_true, y_pred)
    return tnr, fpr, fnr, precision, recall, f1, auc

# Perform k-fold cross-validation
kfold = KFold(n_splits=3, shuffle=True, random_state=42)

all_metrics = []
for train_idx, test_idx in kfold.split(data):
    data_train_fold, data_val_fold = data[train_idx], data[test_idx]

    # Train the model
    model.fit(data_train_fold, data_train_fold, epochs=50, batch_size=32, verbose=0)

    # Make predictions on the validation set
    val_predictions = model.predict(data_val_fold)

    # Binarize predictions using a threshold (e.g., 0.5)
    # threshold = 0.5
    threshold = 0
    val_predictions_binarized = (val_predictions > threshold).astype(int)

    # Calculate performance metrics for this fold
    tnr, fpr, fnr, precision, recall, f1, auc = calculate_metrics(data_val_fold.flatten(), val_predictions_binarized.flatten())

    all_metrics.append((tnr, fpr, fnr, precision, recall, f1, auc))

# Calculate average performance metrics over all folds
mean_metrics = np.mean(all_metrics, axis=0)

# Print the results
print("Average TNR: ", mean_metrics[0])
print("Average FPR: ", mean_metrics[1])
print("Average FNR: ", mean_metrics[2])
print("Average Precision: ", mean_metrics[3])
print("Average Recall: ", mean_metrics[4])
print("Average F1-score: ", mean_metrics[5])
print("Average AUC: ", mean_metrics[6])
```
