```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.ensemble import IsolationForest
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

homes = ['A', 'B', 'C']
data_types = ['sim', 'real']
base_path = "C:/Users/olufe/projects/Journal/dataset/"

# Initialize plot
plt.figure(figsize=(5, 3))

for home in homes:
    for data_type in data_types:
        # Load and preprocess data
        train_data_path = f"{base_path}psa_journal_home_{home}_unsupervise_train.csv"
        test_data_path = f"{base_path}psa_journal_home_{home}_{data_type}_test.csv"

        train_data_df = pd.read_csv(train_data_path)
        test_data_df = pd.read_csv(test_data_path)

        X_train = train_data_df.drop(columns=["Label"])
        y_train = train_data_df["Label"]
        X_test = test_data_df.drop(columns=["Label"])
        y_test = test_data_df["Label"]

        # Normalize data
        scaler = MinMaxScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        # Train model
        isolation_forest = IsolationForest(contamination=0.05, random_state=42)
        isolation_forest.fit(X_train_scaled)

        # Predict outliers
        outliers_test = isolation_forest.predict(X_test_scaled)
        outliers_test = np.where(outliers_test == -1, 1, 0)

        # Compute ROC AUC score and curve
        roc_auc = roc_auc_score(y_test, outliers_test)
        fpr_curve, tpr_curve, _ = roc_curve(y_test, outliers_test)

        # Plot ROC curve
        label = f'Home {home} {data_type.capitalize()} (AUC = {roc_auc:.2f})'
        plt.plot(fpr_curve, tpr_curve, label=label)

plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate', fontsize=10, fontweight='bold')
plt.ylabel('True Positive Rate', fontsize=10, fontweight='bold')
plt.title('ROC Curve for Isolation Forest')
plt.legend(loc='lower right', fontsize=8)
plt.savefig(r'C:\Users\olufe\projects\Journal\Diagrams\IF_roc_all_homes.png', bbox_inches='tight')
plt.show()
```
