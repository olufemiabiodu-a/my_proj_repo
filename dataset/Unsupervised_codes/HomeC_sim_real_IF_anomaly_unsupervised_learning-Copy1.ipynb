```python
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report

# Load your data
file_path = "C:/Users/olufe/projects/Journal/dataset/HomeC_unsupervise_sim_combined_shuffled.csv"
data = pd.read_csv(file_path)

# Separate features (X) and labels (y)
X = data.drop(columns=["Label"])

# Create and fit the Isolation Forest model
isolation_forest = IsolationForest(contamination=0.05, random_state=42)
isolation_forest.fit(X)

# Predict anomalies (1 for anomalies, -1 for normal instances)
predictions = isolation_forest.predict(X)

# Convert -1 to 0 for non-anomalies and 1 for anomalies
predictions[predictions == -1] = 1
predictions[predictions == 1] = 0

# Evaluate the performance using classification report
print(classification_report(data["Label"], predictions))
```
