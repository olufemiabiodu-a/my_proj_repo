## Confusion Matrix_Regeneration

```python
import seaborn as sn
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Original confusion matrix
array = [[6690, 331],
         [533, 6103]]

# Normalize the confusion matrix
array_normalized = np.array(array, dtype=float)
for row in range(len(array)):
    row_sum = np.sum(array[row])
    array_normalized[row] = array[row] / row_sum

plt.rcParams.update({'font.size': 10})
df_cm = pd.DataFrame(array_normalized, range(2), range(2))
LABELS = ['Benign', 'Attack']

plt.figure(figsize=(4, 3))
sn.heatmap(df_cm, xticklabels=LABELS, yticklabels=LABELS, annot=True, fmt=".2%", annot_kws={"size": 10},
           linewidths=0.5, square=True, cbar_kws={'label': 'Home C_real'}, cmap=plt.cm.Greys)

plt.title("Normalized Confusion Matrix - IF")
plt.ylabel('True class', fontsize=10, fontweight='bold')
plt.xlabel('Predicted class', fontsize=10, fontweight='bold')
plt.savefig(r'C:\Users\olufe\projects\Journal\Diagrams\Home_A\IF_cm_sim_normalized.png', bbox_inches='tight')
plt.show()
```
